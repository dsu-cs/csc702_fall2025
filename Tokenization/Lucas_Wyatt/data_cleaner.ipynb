```python
def clean_input(input_file, output_file):
    with open(input_file, "r", encoding="utf-8") as f_in, open(output_file, "w", encoding="utf-8") as f_out:
        for line in f_in:
            # split by quotes
            parts = line.strip().split('"')
            if len(parts) >= 6:
                dialogue = parts[5] # 3rd quoted part
                f_out.write(dialogue + '\n')
```

```python
clean_input('episodeIV.txt', 'episodeIV_cleaned.txt')
clean_input('episodeV.txt', 'episodeV_cleaned.txt')
clean_input('episodeVI.txt', 'episodeVI_cleaned.txt')
```
