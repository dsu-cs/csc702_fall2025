# The goal of this project is to do simple word embeddings through Word2Vec. We were meant to put a small twist on it, so I decided to use the text from the first harry potter book as input. This notebook loads harrypotter.csv containing sentences from the first harry potter book, preprocesses the text, and trains a Word2Vec model to learn semantic relationships between words. Using the trained embeddings, it allows users to type in a word and outputs the most similar words. The model uses skip-gram architecture for better semantic capture.

```python
import pandas as pd
from gensim.models import Word2Vec
from gensim.utils import simple_preprocess
```

```python
df = pd.read_csv("/kaggle/input/harry-potter/harrypotter.csv")
sentences = df.iloc[:, 1].dropna().tolist()
```

```python
tokenized_sentences = [simple_preprocess(sentence) for sentence in sentences]
```

```python
model = Word2Vec(
    sentences=tokenized_sentences,
    vector_size=100,
    epochs=5,
    window=6,
    min_count=2,
    workers=4,
    sg=1
)
```

```python
def find_similar(word, topn=5):
    if word in model.wv:
        similar_words = model.wv.most_similar(word, topn=topn)
        print(f"\nWords most similar to '{word}':")
        for w, score in similar_words:
            print(f"  {w} ({score:.4f})")
    else:
        print(f"\nWord '{word}' not found in vocabulary.")
```

```python
while True:
    user_input = input("\nType a word to find similar ones (or 'exit' to quit): ").strip().lower()
    if user_input == "exit":
        break
    find_similar(user_input)
```
