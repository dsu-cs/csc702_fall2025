```python
# -------------------------------
# Step 1: Imports
# -------------------------------
from faster_whisper import WhisperModel
from transformers import pipeline

import torch
import numpy as np

_original_torch_load = torch.load

def bark_compatible_load(*args, **kwargs):
    kwargs["weights_only"] = False
    return _original_torch_load(*args, **kwargs)

torch.load = bark_compatible_load

from bark import SAMPLE_RATE, generate_audio
from bark.generation import preload_models

from moviepy.editor import AudioFileClip, ImageClip, concatenate_videoclips
from google.colab import files

```

```python

import numpy as np
import transformers
from transformers import pipeline

print("Torch:", torch.__version__)
print("CUDA available:", torch.cuda.is_available())
print("NumPy:", np.__version__)
print("Transformers:", transformers.__version__)
```

```python

# -------------------------------
# Step 2: Upload .m4a file
# -------------------------------
uploaded = files.upload()
audio_path = list(uploaded.keys())[0]
print(f"Uploaded file: {audio_path}")
```

```python
# -------------------------------
# Step 3: Transcribe with Whisper
# -------------------------------
DEVICE = "cuda" if torch.cuda.is_available() else "cpu"
MODEL_SIZE = "medium"  # can be tiny, base, small, medium, large

print("Loading Whisper model...")
whisper_model = WhisperModel(MODEL_SIZE, device=DEVICE)

print("Transcribing audio...")
segments, info = whisper_model.transcribe(audio_path)
lines = [segment.text.strip() for segment in segments if segment.text.strip()]

print("Transcription complete:")
for i, line in enumerate(lines):
    print(f"{i+1}: {line}")
```

```python
# -------------------------------
# Step 4: Detect emotion per line
# -------------------------------
print("\nLoading emotion classifier...")
emotion_classifier = pipeline("text-classification", model="j-hartmann/emotion-english-distilroberta-base", device=0)

line_emotions = []
for line in lines:
    result = emotion_classifier(line)[0]
    line_emotions.append({
        "text": line,
        "emotion": result["label"]
    })

print("Detected emotions:")
for item in line_emotions:
    print(f"{item['text']} --> {item['emotion']}")
```

```python

# -------------------------------
# Step 5: Flip emotions (simple mapping)
# -------------------------------
emotion_map = {
    "joy": "sadness",
    "sadness": "joy",
    "anger": "fear",
    "fear": "anger",
    "surprise": "neutral",
    "neutral": "joy",
    "disgust": "joy"
}

for item in line_emotions:
    item["flipped_emotion"] = emotion_map.get(item["emotion"], "neutral")

print("\nFlipped emotions:")
for item in line_emotions:
    print(f"{item['text']} --> {item['flipped_emotion']}")
```

```python
# -------------------------------
# Step 6: Preload Bark models
# -------------------------------
print("\nPreloading Bark models on GPU...")
preload_models(
    text_use_gpu=True,
    text_use_small=True,
    coarse_use_gpu=True,
    coarse_use_small=True,
    fine_use_gpu=True,
    fine_use_small=True,
    codec_use_gpu=True,
    force_reload=True
)
```

```python
# -------------------------------
# Step 7: Generate new Bark audio
# -------------------------------
import soundfile as sf
audio_segments = []
for i, item in enumerate(line_emotions):
    prompt = f"[{item['flipped_emotion']}] {item['text']}"
    print(f"Generating Bark audio {i+1}/{len(line_emotions)}...")
    audio_array = generate_audio(prompt)
    audio_segments.append(audio_array)

# Concatenate audio segments
final_audio = np.concatenate(audio_segments)
output_wav = "flipped_emotions.wav"

sf.write(output_wav, final_audio, SAMPLE_RATE)
print(f"\nGenerated audio saved to {output_wav}")
```

```python
# -------------------------------
# Step 8: Create cartoon panda video (optimized)
# -------------------------------

import numpy as np
import librosa
from moviepy.editor import ImageClip, AudioFileClip

# Configuration
AUDIO_FILE = "/content/flipped_emotions.wav"  # final Bark audio file
IMAGE_FILE = "/content/Panda.png"
FPS = 30  # video frames per second

# Load audio
audio, sr = librosa.load(AUDIO_FILE, sr=None)
duration = librosa.get_duration(y=audio, sr=sr)

# Calculate energy for lip sync
hop_length = int(sr / FPS)
energy = librosa.feature.rms(y=audio, hop_length=hop_length)[0]

# Ensure energy matches number of frames
num_frames = int(duration * FPS)
if len(energy) < num_frames:
    energy = np.pad(energy, (0, num_frames - len(energy)))
else:
    energy = energy[:num_frames]

# Create base clip
clip = ImageClip(IMAGE_FILE).set_duration(duration)

# Function to scale image for mouth movement
def animate_mouth(get_frame, t):
    frame_idx = min(int(t * FPS), num_frames - 1)
    mouth_open = energy[frame_idx] > 0.02
    scale = 1.0 + 0.05 * mouth_open
    frame = get_frame(t)

    # Ensure frame is 3-channel
    if frame.ndim == 2:  # grayscale
        frame = np.stack([frame]*3, axis=-1)
    elif frame.shape[2] == 1:  # single channel
        frame = np.repeat(frame, 3, axis=2)

    h, w, _ = frame.shape
    import cv2
    frame = cv2.resize(frame, (int(w*scale), int(h*scale)))

    # Center crop/pad to original size
    start_x = (frame.shape[1] - w) // 2
    start_y = (frame.shape[0] - h) // 2
    frame = frame[start_y:start_y+h, start_x:start_x+w]
    return frame

# Apply animation
animated_clip = clip.fl(animate_mouth, apply_to=['mask', 'video'])

# Set audio
animated_clip = animated_clip.set_audio(AudioFileClip(AUDIO_FILE))

# Write out final video
OUTPUT_VIDEO = "panda_video.mp4"
animated_clip.write_videofile(OUTPUT_VIDEO, fps=FPS)

print(f"Video saved as {OUTPUT_VIDEO}")
```
