```python
import torch
from diffusers import DiffusionPipeline

# switch to "mps" for apple devices
pipe = DiffusionPipeline.from_pretrained("stable-diffusion-v1-5/stable-diffusion-v1-5", dtype=torch.bfloat16, device_map="cuda")

```

```python
prompt = "Baseball player hitting a ball"
image = pipe(prompt).images[0]
```

```python
image.save("test.png")
```

```python
guidance_scales = [7.5]
steps = [300, 500, 700]
```

```python
for g in guidance_scales:
    for s in steps:
        image = pipe(prompt, guidance_scale=g, num_inference_steps=s).images[0]
        image.save(f"test_g{g}_s{s}.png")
```

```python
prompt = "Baseball hitter in batter's box, bat on shoulder, looking at pitcher"
```

```python
guidance_scales = [15, 20, 25]
steps = [20, 40, 60]
```

```python
for g in guidance_scales:
    for s in steps:
        image = pipe(prompt, guidance_scale=g, num_inference_steps=s).images[0]
        image.save(f"test_g{g}_s{s}.png")
```

```python
image = Image.open("test_g25_s60.png")
image
```

```python
negative_prompt = "glove, catcher"
for g in guidance_scales:
    for s in steps:
        image = pipe(prompt, negative_prompt=negative_prompt, guidance_scale=g, num_inference_steps=s).images[0]
        image.save(f"test_g{g}_s{s}.png")
```
